```python
#定義一個函式
def sayHello():
    print("Hello, World!")

sayHello()
```

```python
#定義一個函式,有一參數
def say_hello(name):
    print("Hello!", name)


say_hello("Robert")
```

```python
n = 10
n = n + 5
print(n)
```

```python
#複合指定運算子
n = 10
n *= n + 5
print(n)
```

```python
#定義一個函式,輸入一個整數，回傳是否為偶數（True/False）
def is_even(n):
    return n % 2 == 0

#測試函式
print(is_even(4))  # True
print(is_even(5))  # False
```
